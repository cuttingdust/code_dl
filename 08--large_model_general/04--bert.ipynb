```python
from pathlib import Path

model_path = (Path.cwd().parent / "PretrainedModel" / "bert-base-chinese").resolve()
print(f"model_path: {model_path}")

from transformers import BertTokenizer, BertForSequenceClassification
```

```python
tokenizer = BertTokenizer.from_pretrained(model_path)
model = BertForSequenceClassification.from_pretrained(model_path, num_labels=5)
```

```python
tokenizer
```

```python
model
```

# 分词器

```python
# 直接分词
tokenizer.tokenize("2026年欢迎来到黑马学习大模型")
```

```python
# 直接分词
tokenizer("2026年欢迎来到黑马学习大模型")
```

```python
# 句子对处理
print(tokenizer(text="2026年欢迎来到黑马学习大模型", text_pair="新年好"))
print("-" * 100)
print(tokenizer("2026年欢迎来到黑马学习大模型", "新年好"))
```

```python
# 多个句子分词，返回pytorch 的tensor
tokenizer(["2026年欢迎来到黑马学习大模型", "AI算法"], return_tensors="pt", padding=True, truncation=True,
          max_length=None, )
```

# 模型推理计算

```python
inputs = "hello world!"

inputs = tokenizer(inputs, return_tensors="pt")
```

```python
output = model(**inputs)
```

```python
output
```

```python
logits = output.logits
```

```python
# 取概率最大的标签，就是分类标签
logits.argmax()
```

# MLM 预测任务

```python
from transformers import BertForMaskedLM

mlm = BertForMaskedLM.from_pretrained(model_path)
```

```python
mlm
```

```python
mlm_inputs = "下面是一条[MASK][MASK]的新闻，吴京是中国有名的武打明星"
mlm_inputs = tokenizer(mlm_inputs, return_tensors="pt")
mlm_inputs
```

```python
output = mlm(**mlm_inputs)
logits = output.logits
```

```python
logits.shape
```

```python
index = 1  # [MASK] 位置在1 索引
logits[:, index, :]
```

```python
mlm_input_id = logits[:, index, :].argmax()
mlm_input_id
```

```python
# "[MASK]京是中国的首都"
# [MASK] ===> 北
tokenizer.decode(mlm_input_id)
```
